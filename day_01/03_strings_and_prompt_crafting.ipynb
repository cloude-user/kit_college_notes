### ✂️ Step 1: String Indexing & Slicing Visualized

Every string in Python is an ordered sequence of characters.
We can grab any letter using its position:
- **Positive Indexing**: Starts from `0` at the beginning (`0, 1, 2, ...`).
- **Negative Indexing**: Starts from `-1` at the very end (`-1, -2, -3, ...`)!

Take a look at this handy visual guide:

![String Slicing Guide](../images/string_slicing_guide.png)

Let's test this in code:


```python
word = "PYTHON"

print("First letter (word[0]): ", word[0])   # 'P'
print("Last letter  (word[-1]):", word[-1])  # 'N'
print("Second to last (word[-2]):", word[-2])# 'O'

# Slicing: word[start : stop] (Up to, but not including stop)
print("First 3 letters (word[0:3]):", word[0:3]) # 'PYT'
print("From index 2 to end (word[2:]):", word[2:]) # 'THON'
```

### 🧹 Step 2: Essential Text Cleaning Methods

Raw user prompts and web scrape data are often messy—filled with extra spaces, mixed uppercase/lowercase, and weird punctuation.
Python gives us superpowers to clean text in seconds:


```python
messy_input = "   \n  PROMPT: Tell me about Neural Networks???   \n"

# 1. strip() removes leading and trailing whitespace/newlines:
cleaned = messy_input.strip()
print("After strip():\n", repr(cleaned))

# 2. lower() normalizes to lowercase:
lowercase = cleaned.lower()
print("\nAfter lower():\n", lowercase)

# 3. replace() swaps text:
standardized = lowercase.replace("???", ".").replace("prompt: ", "")
print("\nAfter replace():\n", standardized)

# 4. split() splits into a list of words:
words = standardized.split()
print("\nWords list:", words)
print("Total words:", len(words))
```

### ⚡ Step 3: Fast String Joining (Avoid the `+=` Trap!)

Here is an important pro-tip I want you to know:
Because strings in Python are **immutable**, concatenating strings in a loop using `+=` is slow because Python has to create a brand new string on every single step.

The fast, Pythonic way is to collect words in a list and use `''.join()`:


```python
tokens = ["Artificial", "Intelligence", "is", "amazing"]

# Join all tokens with a space between each:
sentence = " ".join(tokens)
print("Joined Sentence:", sentence)

# You can join with hyphens, commas, or anything you like:
tags = ", ".join(tokens)
print("Joined Tags:    ", tags)
```

### 🎨 Step 4: Crafting Dynamic Prompts with F-Strings

F-strings (formatted string literals) allow you to inject any variable directly inside curly braces `{}`.

Even better: What if your AI prompt needs to return a **JSON Schema** that also uses curly braces?
**The trick**: Just double the curly braces: `{{` and `}}`!

Let's see this in action:


```python
def make_ai_prompt(user_topic: str, max_points: int) -> str:
    prompt = f"""### SYSTEM INSTRUCTION
You are an expert AI tutor. Explain the user topic clearly.

### USER TOPIC
{user_topic}

### OUTPUT FORMAT
Respond strictly in JSON with at most {max_points} bullet points:
{{
    "topic": "{user_topic}",
    "bullets": [
        "point 1",
        "point 2"
    ]
}}
"""
    return prompt

# Let's generate a prompt:
my_prompt = make_ai_prompt(user_topic="Transformers in NLP", max_points=3)
print(my_prompt)
```
