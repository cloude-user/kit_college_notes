In this notebook, we look at Python's powerhouse **Collections**.
Whenever you work with AI embeddings, model configurations, conversation histories, or datasets, you will use these four friendly structures:
1. **Lists**: Your flexible, ordered shopping cart.
2. **Tuples**: The locked, immutable list (perfect for fixed dimensions).
3. **Dictionaries**: Labeled key-value pairs (the foundation of AI JSON payloads).
4. **Sets**: The magic deduplicator (guarantees unique items and lightning-fast search).

Let's compare them visually!


### 📊 Lists vs Dictionaries: When Do We Use Which?

Take a look at this comparison:

![Lists vs Dictionaries](../images/list_vs_dictionary.png)

- **Use a List** when order matters and items are accessed by number (`0, 1, 2, ...`). Example: A list of words, a list of embedding numbers.
- **Use a Dictionary** when each piece of information has a specific name or label. Example: `{"role": "user", "content": "Hello"}`.

Let's test both!


```python
# 1. Lists: Flexible, ordered sequences
models = ["gpt-4o", "claude-3-5", "gemini-1.5"]
print("First model in list:", models[0])

# We can append new items:
models.append("llama-3")
print("After append:        ", models)
print("Total models:        ", len(models))
```

### 🔒 Step 2: Tuples (The Lockbox)

A **Tuple** looks like a list, but it is created with round parentheses `(...)` instead of square brackets `[...]`.
The big difference: **Tuples cannot be changed after creation** (they are immutable)!

Why is this useful?
It protects fixed settings, coordinate spaces, and dimensions from accidental changes:


```python
# Tuples: Immutable, fixed records
resolution = (1920, 1080)
model_shape = (512, 768)

print("Resolution width:", resolution[0])

# Try modifying a tuple:
resolution[0] = 3840
# except TypeError as err:
#     print("\n[SAFETY SHIELD] Python stopped us:", err)
#     print("Tuples are locked and cannot be changed!")
```

### 📖 Step 3: Dictionaries (Labeled Key-Value Pairs)

In AI development, almost every interaction with an API returns a dictionary (JSON structure).
Let's build a chat message dictionary and safely inspect its contents:


```python
# A dictionary stores data in key: value pairs:
chat_message = {
    "role": "assistant",
    "content": "Deep learning models learn patterns from training data.",
    "tokens": 12,
    "model": "gpt-4o"
}

# Accessing values by their key:
print("Message Role:   ", chat_message["role"])
print("Message Content:", chat_message["content"])

# Pro-Tip: Always use .get() for safe lookups!
# If a key doesn't exist, .get() returns a default fallback instead of crashing:
status = chat_message.get("finish_reason", "normal_stop")
print("Finish Reason:  ", status)
```

### 🌟 Step 4: Sets (Instant Deduplication & Fast Lookups)

A **Set** is an unordered collection of **unique** elements.
If you give a set duplicate items, it automatically cleans them up!

Checking if an item is in a set takes $O(1)$ constant time (instant), no matter how big the set is!


```python
# In NLP, we often want the unique vocabulary across documents:
raw_tokens = ["attention", "is", "all", "you", "need", "attention", "models", "need", "attention"]

# Convert list to set:
unique_vocab = set(raw_tokens)

print("Original token count: ", len(raw_tokens))
print("Unique vocabulary size:", len(unique_vocab))
print("Unique words:          ", unique_vocab)

# Fast membership check:
print("Is 'attention' in vocabulary?", "attention" in unique_vocab)
```

---
### 🚀 Step 5: List Comprehensions (Writing Clean 1-Line Pipelines)

In JavaScript, you often use `.map()` and `.filter()`.
In Python, we use **List Comprehensions**—they are fast, clean, and loved by Python developers everywhere!

Formula: `[expression for item in iterable if condition]`

Let's see how easy it is to normalize embedding scores:


```python
raw_scores = [0.12, 0.85, 0.94, 0.33, 0.78]

# Keep only high confidence scores (> 0.70) and format as percentages:
high_confidence = [f"{score * 100:.1f}%" for score in raw_scores if score > 0.70]

print("High Confidence Scores:", high_confidence)
```
