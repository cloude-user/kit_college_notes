### 🏷️ Step 1: Variables are Sticky Name Tags (Not Fixed Boxes!)

In languages like C or Java, a variable is like a fixed metal box with a fixed type etched into it.
In Python, variables are **friendly sticky name tags** pointing to objects stored in computer memory!

Take a look at this illustration:

![Variables in Memory](../images/variables_and_memory.png)

When we write:
```python
user_name = "Sundeep"
```
Python creates the string object `"Sundeep"` on the heap, and places the sticky label `user_name` on it!

Let's test this in Python:


````python
# Creating variables is as simple as: name = value
user_name = "Sundeep"
batch_size = 32
learning_rate = 0.001
is_model_ready = True

print("User Name:    ", user_name)
print("Batch Size:   ", batch_size)
print("Learning Rate:", learning_rate)
print("Is Ready:     ", is_model_ready)
````

### 🔢 Step 2: The Core Primitive Data Types

Python has a clean family of built-in types. Let's see them in action:

![Core Data Types](../images/python_core_types.png)

Let's use the built-in `type()` function to inspect the type of any variable:


````python
# Let's inspect the types Python automatically assigns:
tokens = 1500               # Whole number -> int
temperature = 0.7           # Decimal number -> float
system_prompt = "Be helpful" # Text -> str
has_converged = False       # True/False -> bool
missing_data = None         # Absence of value -> NoneType

print("tokens is:        ", type(tokens))
print("temperature is:   ", type(temperature))
print("system_prompt is: ", type(system_prompt))
print("has_converged is: ", type(has_converged))
print("missing_data is:  ", type(missing_data))
````

### ➕ Step 3: Math & Arithmetic in Python

Python is amazing at math. Let's look at the arithmetic operators we use constantly in AI calculations:
- `+` Addition
- `-` Subtraction
- `*` Multiplication
- `/` True division (always gives a float decimal!)
- `//` Floor division (chops off decimals)
- `**` Exponentiation (powers, like $2^{10}$)
- `%` Modulo (remainder)

Let's calculate an LLM API compute cost together:


````python
# Let's compute token costs together:
total_tokens = 75000
cost_per_million = 5.00  # $5.00 per 1,000,000 tokens

# Calculate cost:
cost = (total_tokens / 1_000_000) * cost_per_million

# Notice the nice f-string formatting with 4 decimal places:
print(f"Total Tokens: {total_tokens:,}")
print(f"Estimated Cost: ${cost:.4f}")

# Let's test powers:
print("2 to the power of 8 (2**8):", 2 ** 8)
````

### 🔍 Step 4: Peeking at Memory with `id()`

Because variables are pointers, two names can point to the exact same object in memory!
We can use `id(variable)` to see the actual memory address:

Let's try it:


````python
a = 100
b = a

print("Memory address of a:", id(a))
print("Memory address of b:", id(b))
print("Do 'a' and 'b' share the exact same memory address?", a is b)
````

Now watch what happens when we reassign `a` to a new number:


````python
# Integers are immutable! Reassigning creates a new object:
a = a + 1
print("New memory address of a:", id(a))
print("Old memory address of b:", id(b))
print("Are they still the same?", a is b)
````

### 🦭 Step 5: The Walrus Operator (`:=`)

In Python 3.8+, we have the Walrus Operator (`:=`), so named because it looks like eyes and tusks!
It allows us to **calculate and assign a variable directly inside a check**:

Let's see why this is so helpful for checking token limits:


````python
prompt = "Explain quantum computing in simple terms for a high school student."

# Without walrus, you have to count tokens first, then check.
# With walrus, we do both in one clean line:
if (word_count := len(prompt.split())) > 5:
    print(f"Prompt is detailed enough ({word_count} words). Ready to send to AI!")
else:
    print(f"Prompt is too short ({word_count} words). Please add more context.")
````
