### 🔍 Step 1: How Does Python Actually Run Our Code?

Before we run code, let's understand what happens when you press **Run** on a Python file.

Take a look at this diagram I put together for you:

![How Python Runs Your Code](../images/python_execution_model.png)

Here is the simple flow:
- **Step 1**: You and I write clean, human-readable Python code in a `.py` or `.ipynb` file.
- **Step 2**: Python's compiler quietly translates our text into compact, platform-independent **Bytecode** (`.pyc`).
- **Step 3**: The **Python Virtual Machine (PVM)** executes those bytecode instructions on your computer.

### 📝 Step 2: Comments (Leaving Notes for Yourself and Me)
In Python, any line starting with a hash symbol (`#`) is a **comment**.
Python completely ignores comments when running code. We use them to explain our thinking, document prompt ideas, or leave helpful hints.

Run the cell below:


```python
# This is a single-line comment. Python will ignore this line!
# Let's define our AI model name:
model_name = "gemini-1.5-pro"

# Now let's print it:
print("Our active AI model is:", model_name)
```

You can also write multi-line comments or docstrings using triple quotes (`"""` or `'''`):


```python
"""
This is a multi-line docstring!
We often use these at the top of AI functions to explain
what inputs the function expects and what output it returns.
"""
print("Docstrings are great for explaining complex AI prompts!")
```

### 📐 Step 3: Indentation (No Curly Braces Needed!)

If you have used JavaScript, Java, or C++, you are probably used to wrapping blocks of code in curly braces `{ ... }` and ending lines with semicolons `;`.

In Python:
1. **No semicolons needed** at the end of lines!
2. **Indentation (4 spaces)** defines code blocks!
3. Block headers (like `if`, `for`, `def`) always end with a colon (`:`).

Let's see how clean and readable this looks:


```python
# Let's check a simple condition using Python's clean indentation:
confidence_score = 0.95

if confidence_score > 0.80:
    # Notice the 4-space indent here! This line only runs if the condition is True:
    print("High confidence! The AI is very sure of this answer.")
    print(f"Confidence value: {confidence_score * 100}%")

print("This line is NOT indented, so it runs no matter what!")
```
