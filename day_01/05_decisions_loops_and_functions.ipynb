### 🔄 Step 2: Loops in Python (`for` loops made friendly)

In Python, loops don't need tricky index counters like `for (let i = 0; i < len; i++)`.
You just loop over the items directly: `for item in items:`

And if you ever DO want the item index, just use `enumerate()`!


```python
retrieved_docs = [
    "Transformers rely on multi-head attention.",
    "RAG fetches external knowledge from vector databases.",
    "Fine-tuning adapts weights to specific task domains."
]

print("Looping over documents with enumerate():\n")
for index, doc in enumerate(retrieved_docs, start=3):
    print(f"[{index}] {doc}")
```

### 🛠️ Step 4: Writing Your Own Reusable Functions with `def`

Functions are our building blocks. They take inputs (parameters), do useful work, and return an output.
Let's build a clean function that evaluates whether a text prompt is safe and well-formatted:


```python
def inspect_prompt(prompt_text: str, max_allowed_words: int = 50) -> dict:
    """
    Inspects user prompt length and returns a summary status.
    """
    words = prompt_text.strip().split()
    word_count = len(words)
    
    is_safe = word_count <= max_allowed_words
    
    return {
        "word_count": word_count,
        "is_safe_length": is_safe,
        "first_word": words[0] if words else "",
        "status": "APPROVED" if is_safe else "EXCEEDS_LIMIT"
    }

# Let's test our function:
result1 = inspect_prompt("Explain neural networks in simple terms.")
print("Test 1 Result:", result1)

result2 = inspect_prompt("word " * 60)
print("Test 2 Result:", result2)
```

### 🤖 Step 5: Our First Mini AI Agent Dispatcher

Let's combine everything we have learned—strings, dictionaries, decisions, and functions—into a simple AI Agent tool dispatcher!


```python
def simple_ai_agent(user_query: str) -> str:
    query_clean = user_query.strip().lower()
    
    # Simple intent routing:
    if "hello" in query_clean or "hi" in query_clean:
        return "Hello! I am your AI assistant. How can I help you today?"
        
    elif "math" in query_clean or "calculate" in query_clean:
        return "I can help with math! 2 + 2 = 4, and 2**10 = 1024."
        
    elif "model" in query_clean:
        return "I am currently powered by Python Foundations 1.0!"
        
    else:
        return f"I have received your query: '{user_query}'. Synthesizing response..."

# Let's talk to our agent:
print(simple_ai_agent("Hi there!"))
print(simple_ai_agent("Can you do some math for me?"))
print(simple_ai_agent("Which model are you?"))
print(simple_ai_agent("Explain quantum computing"))
```
