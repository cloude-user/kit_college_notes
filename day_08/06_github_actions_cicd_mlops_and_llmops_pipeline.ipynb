## 1. Production GitHub Actions CI/CD Pipeline
Every git push triggers automated validation:



```python
github_actions_workflow = '''name: Production AI CI/CD Pipeline

on:
  push:
    branches: [main, develop]
  pull_request:
    branches: [main]

jobs:
  quality_and_tests:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout Codebase
        uses: actions/checkout@v4

      - name: Set up Python 3.12
        uses: actions/setup-python@v5
        with:
          python-version: "3.12"
          cache: "pip"

      - name: Install Dependencies
        run: |
          pip install --upgrade pip
          pip install -r requirements.txt ruff mypy pytest pytest-cov

      - name: Code Quality & Linting (Ruff)
        run: ruff check .

      - name: Strict Type Checking (Mypy)
        run: mypy --ignore-missing-imports src/

      - name: Run Automated Pytest Suite with Coverage
        run: pytest tests/ --cov=src --cov-fail-under=80

  docker_build_and_deploy:
    needs: quality_and_tests
    if: github.ref == 'refs/heads/main'
    runs-on: ubuntu-latest
    steps:
      - name: Checkout Codebase
        uses: actions/checkout@v4

      - name: Build Production Docker Image
        run: docker build -t my-ai-service:${{ github.sha }} .
'''

print("=== GITHUB ACTIONS WORKFLOW YAML ===")
print(github_actions_workflow)
```

## 2. MLOps vs. LLMOps Comparison
| Dimension | Traditional MLOps | Modern LLMOps |
| :--- | :--- | :--- |
| **Model Creation** | Training weights from scratch (scikit-learn, PyTorch) | Foundation models accessed via API or fine-tuned |
| **Primary Metric** | Statistical ROC-AUC, F1-Score, MSE | Faithfulness, Hallucination Rate, Answer Relevance |
| **Data Tracking** | Feature stores, labeled datasets | Vector embeddings, chunk stores, prompt registries |
| **CI/CD Quality Gate** | Model accuracy on test split | Automated Ragas benchmark scores in CI runner |



---
## 3. Implementing an Automated LLMOps Evaluation Gate
Before merging code, evaluate whether RAG faithfulness meets the 90% threshold:



```python
def evaluate_llmops_pipeline(test_cases: List[Dict[str, Any]], min_faithfulness: float = 0.90) -> Dict[str, Any]:
    # Simulates automated LLMOps evaluation gate in CI/CD runner
    passed = 0
    scores = []
    
    for case in test_cases:
        # Check that answer facts are backed by context
        context_words = set(case["context"].lower().split())
        answer_words = set(case["answer"].lower().split())
        overlap = len(context_words.intersection(answer_words)) / (len(answer_words) + 1e-6)
        score = min(1.0, overlap * 1.5)
        scores.append(score)
        if score >= min_faithfulness:
            passed += 1

    avg_score = sum(scores) / len(scores)
    gate_passed = avg_score >= min_faithfulness

    return {
        "total_test_cases": len(test_cases),
        "passed_cases": passed,
        "average_faithfulness_score": round(avg_score, 3),
        "ci_cd_gate_status": "PASSED - READY FOR DEPLOYMENT" if gate_passed else "FAILED - BLOCKED"
    }

sample_eval_cases = [
    {"context": "Dr. Smith prescribed Metformin 500mg BID to Rahul Sharma.", "answer": "Dr. Smith prescribed Metformin 500mg BID to Rahul."},
    {"context": "Hospital protocol mandates HbA1c screening quarterly.", "answer": "Quarterly HbA1c screening is required by hospital protocol."}
]

eval_report = evaluate_llmops_pipeline(sample_eval_cases)
print("=== AUTOMATED LLMOPS CI/CD GATE EVALUATION ===")
print(json.dumps(eval_report, indent=2))
```
