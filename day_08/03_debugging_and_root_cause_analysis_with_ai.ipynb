## 1. The 4-Part Production Debugging Prompt
Never paste a raw error without context. Always provide:
1. **Expected Behavior:** What the code *should* do.
2. **Actual Failure:** The exact error message and full stack trace.
3. **Reproduction Code:** The minimal code snippet triggering the bug.
4. **Environment Context:** Python version, OS, database state, library versions.



````python
debugging_prompt_template = '''DEBUGGING ASSISTANCE REQUEST:
Context: Python 3.12, FastAPI, SQLite, Pydantic v2

Expected Behavior:
Querying patient prescriptions should return a list of prescription objects.

Actual Failure:
pydantic_core._pydantic_core.ValidationError: 1 validation error for PrescriptionRecord
dosage
  Field required [type=missing, input_value={'patient': 'Rahul', 'medication': 'Metformin'}, input_type=dict]

Code Snippet:
```python
class PrescriptionRecord(BaseModel):
    patient: str
    medication: str
    dosage: str

records = [PrescriptionRecord(**row) for row in db_cursor.fetchall()]
```

Task:
1. Explain the root cause of this validation error.
2. Explain how to handle legacy database rows where dosage might be NULL using Pydantic defaults.
3. Provide the corrected, robust code snippet.
'''

print("=== 4-PART DEBUGGING PROMPT TEMPLATE ===")
print(debugging_prompt_template)
````
