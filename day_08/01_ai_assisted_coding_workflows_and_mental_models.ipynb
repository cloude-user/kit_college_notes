## 1. How AI Coding Assistants Work Under the Hood
An AI coding assistant does not "know" your application. It uses a 5-stage pipeline:

```
┌────────────────────────┐      ┌────────────────────────┐      ┌────────────────────────┐
│ 1. Your Prompt & Files │ ───► │  2. Context Collected   │ ───► │ 3. LLM Predicts Code   │
│   (Active tab, cursor) │      │  (Imports, AST, tabs)  │      │  (Token probability)   │
└────────────────────────┘      └────────────────────────┘      └───────────┬────────────┘
                                                                            │
┌────────────────────────┐      ┌────────────────────────┐                  │
│  5. You Review & Commit│ ◄─── │ 4. Suggestion Shown    │ ◄────────────────┘
│   (Developer is pilot) │      │  (Ghost text / diff)   │
└────────────────────────┘      └────────────────────────┘
```

### The 3 Golden Rules:
1. **The Context Rule:** Better context (open relevant tabs, clear type signatures, examples) yields higher quality suggestions.
2. **The Hallucination Rule:** The model can be confidently wrong (inventing library methods or deprecated syntax). Always verify.
3. **The Ownership Rule:** You are the pilot; AI is the co-pilot. You are 100% accountable for the code you commit.



````python
def simulate_context_collection(active_file: str, open_tabs: List[str], max_context_kb: int = 64) -> Dict[str, Any]:
    # Simulates IDE context collection engine
    context_files = [active_file]
    for tab in open_tabs:
        if tab != active_file and len(context_files) < 5:
            context_files.append(tab)
            
    return {
        "primary_file": active_file,
        "included_context_files": context_files,
        "context_window_usage": f"{len(context_files) * 8}KB / {max_context_kb}KB",
        "strategy": "AST_Imports_Plus_Active_Tabs"
    }

ctx = simulate_context_collection("service.py", ["service.py", "models.py", "schema.sql", "test_service.py"])
print("=== IDE CONTEXT ENGINE STATUS ===")
print(json.dumps(ctx, indent=2))
````
