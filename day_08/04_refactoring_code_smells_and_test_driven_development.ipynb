## 1. Refactoring Code Smells: Before vs. After
Look at this messy, monolithic legacy function:



```python
# BEFORE: High cyclomatic complexity, string parsing, no type hints, hardcoded values
def process_patient_data(data):
    if data != None:
        if "name" in data:
            if "age" in data and data["age"] > 18:
                if "prescriptions" in data:
                    res = []
                    for p in data["prescriptions"]:
                        if p["drug"] == "Metformin":
                            res.append(p["dose"])
                    return res
    return []

print("Legacy function defined. Notice the deep nesting (arrow anti-pattern)!")
```

---
## 2. Safe Refactoring Prompt
Let's see how a senior engineer prompts AI to refactor:



```python
refactor_prompt = '''Refactor this function to production standards:
1. Eliminate deep nesting using early returns (guard clauses).
2. Add full Python 3.12 type annotations.
3. Replace dictionary lookups with a validated Pydantic model.
4. Reduce cyclomatic complexity below 3.
5. Provide a Google-style docstring.
'''

# AFTER: Clean, type-safe, self-documenting code
from pydantic import BaseModel
from typing import List, Optional

class Prescription(BaseModel):
    drug: str
    dose: str

class Patient(BaseModel):
    name: str
    age: int
    prescriptions: List[Prescription] = []

def get_adult_metformin_dosages(patient: Optional[Patient]) -> List[str]:
    # Returns Metformin dosages for adult patients using clean guard clauses
    if not patient or patient.age <= 18:
        return []
        
    return [p.dose for p in patient.prescriptions if p.drug.lower() == "metformin"]

print("[+] Refactored version successfully loaded!")
```

---
## 3. Test-Driven Development (TDD) with Pytest
Write the test suite *first* to establish the specification:



```python
def test_adult_patient_with_metformin():
    p = Patient(
        name="Rahul Sharma",
        age=45,
        prescriptions=[Prescription(drug="Metformin", dose="500mg BID"), Prescription(drug="Atorvastatin", dose="20mg")]
    )
    doses = get_adult_metformin_dosages(p)
    assert doses == ["500mg BID"]

def test_minor_patient_returns_empty():
    p = Patient(name="Child", age=12, prescriptions=[Prescription(drug="Metformin", dose="250mg")])
    assert get_adult_metformin_dosages(p) == []

def test_none_patient_returns_empty():
    assert get_adult_metformin_dosages(None) == []

# Run unit tests directly
test_adult_patient_with_metformin()
test_minor_patient_returns_empty()
test_none_patient_returns_empty()
print("[ALL TESTS PASSED] TDD suite verified 100% correctness!")
```
