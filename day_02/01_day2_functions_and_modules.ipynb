### 🛠️ Step 1: Anatomy of a Function Visualized

Think of a function as a friendly, reusable machine in our college campus:
1. You pass it **Inputs** (arguments like marks or student name).
2. It executes its **Process block** (indented 4 spaces).
3. It **Returns** the computed result back to you!

![Function Anatomy](../../images/day2_functions_anatomy.png)

Let's write our first college grade calculator together:


```python
# Defining our first college helper function:
def calculate_gpa(marks: list[float], student_name: str = "KIT Student") -> dict:
    """Calculates average percentage and letter grade for a KIT student."""
    if not marks:
        return {"name": student_name, "percentage": 0.0, "grade": "F"}
        
    avg = sum(marks) / len(marks)
    
    if avg >= 90: grade = "A+"
    elif avg >= 80: grade = "A"
    elif avg >= 70: grade = "B"
    elif avg >= 60: grade = "C"
    else: grade = "F"
    
    return {
        "name": student_name,
        "total_subjects": len(marks),
        "percentage": round(avg, 2),
        "grade": grade
    }

# Let's test our function with sample marks from semester exams:
sundeep_marks = [92.5, 88.0, 95.0, 89.5, 94.0]
result = calculate_gpa(sundeep_marks, student_name="Sundeep (KIT AI)")
print("Semester Result:", result)
```

### 🌟 Step 2: Flexible Parameters with `*args` and `**kwargs`

What if a student takes 3 electives, but another student takes 6 electives?
- `*args` gathers extra positional items into a **tuple**.
- `**kwargs` gathers extra keyword details into a **dictionary**.

Let's see how flexible this makes our campus registration function:


```python
def register_kit_student(name: str, usn: str, block= "A", *electives, **extra_details):
    print(f"\n--- Enrolling Student: {name} ({usn}) ---")
    print("Electives Chosen (tuple via *args):", electives)
    print("Extra Profile Data (dict via **kwargs):", extra_details)
    print("block:",block)
    
    if "scholarship" in extra_details:
        print(f"🎉 Merit Scholarship Applied: {extra_details['scholarship']}% off!")

# Testing with variable electives and keyword details:
register_kit_student(
    "Priya Sharma",
    "1KT23AI042",
    "Deep Learning", "Computer Vision", "Reinforcement Learning",
    hostel="Block B",
    scholarship=25,
    club="AI Robotics Club"
)


register_kit_student(
    "nikith",
    "3vc19cs200"
)
```

### ⏱️ Step 3: Decorators (Adding Superpowers Without Changing Code)

A **Decorator** is like a security guard or timing stopwatch standing outside a college classroom door.
It runs code **before** and **after** your function automatically!

Let's build a `@track_execution_time` decorator:


```python
import time
from functools import wraps

def track_execution_time(func):
    """Stopwatch decorator that measures how many milliseconds a function takes."""
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed_ms = (time.perf_counter() - start) * 1000
        print(f"[@track_execution_time] '{func.__name__}' finished in {elapsed_ms:.3f} ms")
        return result
    return wrapper

# Let's apply our decorator using the @ symbol:
@track_execution_time
def train_mini_ai_model(epochs: int):
    print(f"Training college neural net for {epochs} epochs...")
    time.sleep(0.05)  # Simulate model training
    return "Model converged with 98.4% accuracy!"

print(train_mini_ai_model(epochs=50))
```

### ⚡ Step 4: Quick 1-Line `lambda` Functions

Sometimes you just need a quick, throwaway formula without writing a full `def`.
Python gives us `lambda` expressions!
Syntax: `lambda inputs: output`

Let's rank KIT students by their test scores:


```python
kit_students = [
    {"name": "Ananya", "score": 88},
    {"name": "Vikram", "score": 96},
    {"name": "Rohan",  "score": 79},
    {"name": "Sneha",  "score": 92}
]

# Sort students descending by score using a lambda key:
ranked = sorted(kit_students, key=lambda s: s["score"], reverse=True)

print("🏆 KIT College AI Batch Leaderboard:")
for rank, student in enumerate(ranked, start=1):
    print(f"  Rank {rank}: {student['name']:<10} - Score: {student['score']}")
```
