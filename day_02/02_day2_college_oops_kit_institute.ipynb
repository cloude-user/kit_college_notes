### 🧠 Concept 1: Classes vs Objects & Memory Inspection

- **Class**: The blueprint defined in code. It defines what attributes and behaviors every member of this type will have.
- **Object**: A real instance created in RAM at runtime. It has a unique memory address (`id()`), a type (`type()`), and its own state dictionary (`__dict__`).
- **Class Attribute**: Shared across ALL instances (e.g. `institute_name = "KIT College Institute"`).
- **Instance Attribute**: Unique to each individual object (e.g. `self.name`, `self.usn`).

Let's test this in code:


````python
# 1. Classes vs Objects & Memory Exploration
class CampusEntity:
    # Class-level attribute: Shared by all objects in the campus!
    institute_name = "KIT College Institute"
    affiliating_university = "VTU"

    def __init__(self, name: str):
        # Instance attribute: Unique to each individual entity!
        self.name = name

# Let's instantiate two real campus entities:
member1 = CampusEntity("Sundeep") # new_CampusEntity()
member2 = CampusEntity("Dr. Anand")

print("=== MEMORY & IDENTITY INSPECTION ===")
print(f"member1: {member1.name} | Type: {type(member1)} | Memory Address (id): {hex(id(member1))}")
print(f"member2: {member2.name} | Type: {type(member2)} | Memory Address (id): {hex(id(member2))}")

# Inspect their internal instance dictionary:
print("\nmember1 internal state (__dict__):", member1.__dict__)
print("member2 internal state (__dict__):", member2.__dict__)

# Class attributes are shared:
print(f"\nShared Institute: member1 -> '{member1.institute_name}' | member2 -> '{member2.institute_name}'")
print(f"Same memory reference for class attribute: {member1.institute_name is member2.institute_name}")
````

### ⚖️ Concept 1B: Class Variables vs Instance Variables & The Priority Shadowing Rule

One of the most critical topics for senior developers is understanding **Class Variables** vs **Instance Variables** and mastering **Attribute Lookup Priority**.

#### 1. What is the Difference?
- **Instance Variable (`self.attr`)**:
  - Defined inside methods (typically `__init__`) using `self.attr = value`.
  - Stored inside `instance.__dict__`.
  - Unique to each individual object (e.g. `s1.usn`, `s1.name`, `s1.cgpa`).
- **Class Variable (`ClassName.attr`)**:
  - Defined directly inside the class body, outside any methods.
  - Stored inside `ClassName.__dict__`.
  - Shared across ALL instances (e.g. `institute_code = "KIT_2026"`, `min_passing_gpa = 4.0`).
  - Allocated once in memory.

---

### 🔍 The Attribute Resolution Priority Order (Lookup Hierarchy)
When you write `value = student.attribute`, how does Python find it?
```
1. PRIORITY 1: Look inside instance.__dict__ (The object's own memory)
   -> Found? Returns the instance value immediately! (SHADOWS the class variable!)
2. PRIORITY 2: Look inside Class.__dict__ (The shared blueprint memory)
   -> Found? Returns the shared class value.
3. PRIORITY 3: Look inside Base Classes (traversing the MRO chain)
4. PRIORITY 4: If not found anywhere -> Raises AttributeError!
```

> **CRITICAL RULE: The Assignment vs Lookup Trap!**
> - When **reading** (`s1.institute_code`), Python falls back to the class variable if missing in `s1`.
> - When **writing** (`s1.institute_code = "KIT_AUTONOMOUS"`), Python **NEVER** modifies the class variable!
> - It creates a brand-new instance attribute inside `s1.__dict__`, which **SHADOWS** the class variable for `s1` only! `s2` and `Student` remain untouched!

### 🛠️ Method Taxonomy: Instance Method vs `@classmethod` vs `@staticmethod`
- **Instance Method (`def method(self)`)**:
  - Automatically receives the instance as `self`.
  - Used when you need to read or modify **instance state** (`self.name`, `self.cgpa`).
- **Class Method (`@classmethod def method(cls)`)**:
  - Automatically receives the class as `cls` (not an instance!).
  - Used for **Alternative Constructors** (e.g., `Student.from_csv_row()`) or modifying **shared class state** (`cls.min_passing_gpa = 4.5`).
- **Static Method (`@staticmethod def method()`)**:
  - Receives neither `self` nor `cls`.
  - Behaves like a standard standalone function, but placed inside the class namespace because it is logically related (e.g., `Student.validate_usn(usn)`).

Now let's verify all of this in runnable code!


````python
# 2. Comprehensive Demonstration: Priority, Shadowing, Classmethods & Staticmethods
import re

class KITStudent:
    # --- CLASS-LEVEL VARIABLES (Shared by all instances) ---
    institute_code = "KIT_2026"
    min_passing_gpa = 4.0
    total_enrolled_count = 0
    
    # ⚠️ COMMON TRAP: Never put a mutable list here as a class variable!
    # shared_clubs = []  <-- If someone appends here, EVERY student shares it!

    def __init__(self, name: str, usn: str, cgpa: float):
        # --- INSTANCE-LEVEL VARIABLES (Unique to each student) ---
        self.name = name
        self.usn = usn
        self.cgpa = cgpa
        self.personal_clubs: list[str] = []  # Safe: each instance gets its own list!
        
        # Increment shared class counter:
        KITStudent.total_enrolled_count += 1

    # 1. INSTANCE METHOD: Operates on 'self' (individual student state)
    def update_cgpa(self, new_grade: float):
        self.cgpa = new_grade
        print(f"[INSTANCE METHOD] Updated CGPA for {self.name} ({self.usn}) to {self.cgpa}")

    def join_club(self, club_name: str):
        self.personal_clubs.append(club_name)
        print(f"[INSTANCE METHOD] {self.name} joined club: '{club_name}'")

    # 2. CLASS METHOD: Operates on 'cls' (Class-wide state & Alternative Constructors)
    @classmethod
    def set_minimum_passing_gpa(cls, new_min: float):
        """Modifies the shared class variable for the entire institution!"""
        cls.min_passing_gpa = new_min
        print(f"[CLASS METHOD] Institution-wide min passing GPA updated to: {cls.min_passing_gpa}")

    @classmethod
    def from_csv_row(cls, csv_line: str):
        """Alternative Constructor: Creates a KITStudent directly from a CSV text string!"""
        parts = [p.strip() for p in csv_line.split(",")]
        name, usn, cgpa_str = parts[0], parts[1], parts[2]
        # Validate USN via static method before instantiating:
        if not cls.validate_usn(usn):
            raise ValueError(f"Invalid USN format: {usn}")
        return cls(name=name, usn=usn, cgpa=float(cgpa_str))

    # 3. STATIC METHOD: Pure utility function (Needs neither self nor cls)
    @staticmethod
    def validate_usn(usn: str) -> bool:
        """Validates standard VTU/KIT USN format: e.g. '1KT23AI001'"""
        pattern = r"^1KT\d{2}[A-Z]{2}\d{3}$"
        return bool(re.match(pattern, usn.upper()))

    def __str__(self) -> str:
        return f"Student: {self.name} | USN: {self.usn} | CGPA: {self.cgpa} | Code: {self.institute_code}"

# =============================================================================
# EXPERIMENT 1: ATTRIBUTE RESOLUTION PRIORITY & SHADOWING
# =============================================================================
print("=== EXPERIMENT 1: ATTRIBUTE LOOKUP & SHADOWING ===")
s1 = KITStudent("Sundeep", "1KT23AI001", 9.6)
s2 = KITStudent("Rahul", "1KT23AI002", 8.4)

print(f"s1 institute_code: {s1.institute_code} (Resolved from class: KITStudent.institute_code)")
print(f"s2 institute_code: {s2.institute_code} (Resolved from class: KITStudent.institute_code)")

# Now watch what happens when we assign to s1:
print("\n--- Assigning s1.institute_code = 'KIT_AUTONOMOUS' ---")
s1.institute_code = "KIT_AUTONOMOUS"

print(f"s1 institute_code : '{s1.institute_code}'  <- Priority 1: Read from s1.__dict__ (SHADOWED!)")
print(f"s2 institute_code : '{s2.institute_code}'  <- Priority 2: Still read from class!")
print(f"KITStudent.code   : '{KITStudent.institute_code}'  <- Class variable was NEVER modified!")

print("\nInternal state inspection:")
print("s1.__dict__:", s1.__dict__)  # Contains 'institute_code': 'KIT_AUTONOMOUS'!
print("s2.__dict__:", s2.__dict__)  # Does NOT contain 'institute_code'!

# If we delete the shadowed instance variable from s1:
del s1.institute_code
print("\nAfter 'del s1.institute_code':")
print(f"s1 institute_code : '{s1.institute_code}' (Fell back to class variable again!)")

# =============================================================================
# EXPERIMENT 2: CLASSMETHOD & ALTERNATIVE CONSTRUCTOR
# =============================================================================
print("\n=== EXPERIMENT 2: CLASSMETHODS ===")
# Alternative Constructor parsing a CSV string directly:
csv_record = "Priya Sharma, 1KT23AI042, 9.8"
s3 = KITStudent.from_csv_row(csv_record)
print("Created Student via @classmethod factory:")
print(" ", s3)

# Modifying class-wide state via classmethod:
KITStudent.set_minimum_passing_gpa(4.5)
print(f"Shared Passing GPA for all students: s1 -> {s1.min_passing_gpa} | s3 -> {s3.min_passing_gpa}")
print(f"Total enrolled student counter (Class variable): {KITStudent.total_enrolled_count}")

# =============================================================================
# EXPERIMENT 3: STATICMETHOD FOR VALIDATION
# =============================================================================
print("\n=== EXPERIMENT 3: STATICMETHODS ===")
test_usn_1 = "1KT23AI001"
test_usn_2 = "INVALID_ROLL_123"

print(f"Is '{test_usn_1}' a valid USN? -> {KITStudent.validate_usn(test_usn_1)}")
print(f"Is '{test_usn_2}' a valid USN? -> {KITStudent.validate_usn(test_usn_2)}")
````

### 📋 Real-World Decision Matrix: "When to Use What?"

Here is the exact reference table used by principal software architects:

| Concept | Scope | Syntax | When to Use? | Real-World Campus / AI Example |
| :--- | :--- | :--- | :--- | :--- |
| **Instance Variable** | Per-Object | `self.usn = "1KT23AI001"` | Storing state unique to each individual object. | Student USN, individual CGPA, person name. |
| **Class Variable** | Per-Class (Shared) | `institute_code = "KIT_2026"` | Storing shared configurations, constants, or global counters. | University code, minimum pass mark, total student counter. |
| **Instance Method** | Per-Object | `def enroll(self, course):` | Normal methods that need to read or mutate individual object state. | `s1.enroll("NLP")`, `s1.calculate_gpa()`. |
| **Class Method** | Per-Class | `@classmethod`<br>`def from_csv(cls, row):` | Alternative constructors (factories) or mutating shared class state. | `Student.from_csv(line)`, `Student.from_json(data)`, `Student.set_pass_mark()`. |
| **Static Method** | Isolated Utility | `@staticmethod`<br>`def validate(usn):` | Utility functions that don't touch `self` or `cls`, but belong in class namespace. | `Student.validate_usn("1KT23AI001")`, regex format checkers, pure math helpers. |

> **IITian Rule of Thumb on Priority**:
> Attribute lookup ALWAYS favors `instance.__dict__` first. If you assign `instance.class_var = new_val`, you are NOT modifying the class variable—you are creating a local instance variable that **shadows** it! To modify the shared class variable, you MUST write `ClassName.class_var = new_val` or use a `@classmethod`!

### 🛡️ Concept 2: Abstraction (`abc.ABC` & `@abstractmethod`)

**What is Abstraction?**
Abstraction hides low-level complexity and enforces a strict contract. In our college, the governing board declares:
*"Every single person working or studying on campus MUST have a role summary and a monthly financial payout (stipend, salary, or fee status)!"*

We use Python's built-in `abc` module (`ABC` = Abstract Base Class) and the `@abstractmethod` decorator.

> **Key Rule**: You **CANNOT** instantiate an abstract class directly if it has unimplemented abstract methods. Subclasses **MUST** implement them!


````python
from abc import ABC, abstractmethod

class CollegeMember(ABC):
    """Abstract Base Class representing any human associated with KIT College."""
    institute_name = "KIT College Institute"
    
    def __init__(self, name: str, email: str, campus_id: str):
        self.name = name
        self.email = email
        self.campus_id = campus_id
        
    @abstractmethod
    def get_role_summary(self) -> str:
        """Every derived campus member MUST define their specific role!"""
        pass
        
    @abstractmethod
    def calculate_monthly_payout(self) -> float:
        """Every campus member has a financial formula: salary, stipend, or zero."""
        pass

    def __str__(self) -> str:
        return f"[{self.campus_id}] {self.name} <{self.email}>"

# Let's verify that Python BLOCKS direct instantiation of this abstract class:
try:
    print("Attempting to instantiate abstract base class CollegeMember directly...")
    ghost_member = CollegeMember("Ghost", "ghost@kit.edu", "KIT_000")
except TypeError as error:
    print("\n[ABSTRACTION SHIELD IN ACTION]:")
    print("Caught TypeError as expected:", error)
 
````

### 🔒 Concept 3: Encapsulation & Access Modifiers

In Python, we manage data visibility using naming conventions:
1. **Public** (`self.name`, `self.usn`): Freely accessible anywhere.
2. **Protected** (`self._conduct_score`): Single underscore prefix. Signals to other developers: *"This is internal; treat with care"*.
3. **Private** (`self.__cgpa`, `self.__salary`): Double underscore prefix. Triggers **Name Mangling** so it cannot be accidentally accessed from outside.
4. **`@property` Getter & Setter**: Provides clean pythonic attribute access (`student.cgpa`) while enforcing validation rules (e.g. $0.0 \le \text{CGPA} \le 10.0$).
5. **Python's Philosophy**: *"We are all consenting adults here"*. Python doesn't build unbreakable walls; private variables can still be reached via name mangling (`obj._Student__cgpa`), but doing so violates the API contract.

Let's test encapsulation with our `Student` class:


````python
class Student(CollegeMember):
    """Represents a student enrolled at KIT College with a USN."""
    
    def __init__(self, name: str, email: str, usn: str, branch: str, initial_cgpa: float = 0.0):
        # Initialize the abstract base class attributes:
        CollegeMember.__init__(self, name=name, email=email, campus_id=usn)
        
        self.usn = usn                          # Public attribute (e.g. '1KT23AI001')
        self.branch = branch                    # Public attribute
        self._conduct_score = 100               # Protected attribute (convention: internal)
        self.__cgpa = 0.0                       # Strictly Private attribute!
        self.enrolled_courses: list[str] = []   # Public list
        self.completed_credits = 0              # Public integer for operator overloading
        
        # Use our property setter to validate the initial grade:
        self.cgpa = initial_cgpa

    # --- ENCAPSULATION: GETTER & SETTER ---
    @property
    def cgpa(self) -> float:
        """Getter: Exposes CGPA cleanly as student.cgpa without parentheses."""
        return self.__cgpa

    @cgpa.setter
    def cgpa(self, new_grade: float):
        """Setter: Strictly validates that the grade is between 0.0 and 10.0!"""
        if not isinstance(new_grade, (int, float)):
            raise TypeError(f"CGPA must be a number! Got {type(new_grade).__name__}")
        if 0.0 <= new_grade <= 10.0:
            self.__cgpa = round(float(new_grade), 2)
        else:
            raise ValueError(f"Invalid CGPA {new_grade}! KIT College CGPA must be between 0.0 and 10.0.")

    @cgpa.deleter
    def cgpa(self):
        """Deleter: Safeguard student graduation records from accidental deletion."""
        print(f"[SECURITY ALERT] Resetting CGPA for {self.usn} to 0.0.")
        self.__cgpa = 0.0

    # Implementing abstract methods:
    def get_role_summary(self) -> str:
        return f"Student [USN: {self.usn}] studying {self.branch} (CGPA: {self.__cgpa})"

    def calculate_monthly_payout(self) -> float:
        # Students pay fees, but high achievers get a merit stipend!
        if self.__cgpa >= 9.5:
            return 2500.0  # Monthly merit reward scholarship
        return 0.0

    # Overloading __str__ (Friendly report card) and __repr__ (Developer debug)
    def __str__(self) -> str:
        return f"🎓 KIT Student: {self.name} | USN: {self.usn} | Branch: {self.branch} | CGPA: {self.__cgpa}"

    def __repr__(self) -> str:
        return f"Student(name={self.name!r}, usn={self.usn!r}, branch={self.branch!r}, cgpa={self.__cgpa})"

# Let's instantiate Sundeep at KIT College!
sundeep = Student(
    name="Sundeep",
    email="sundeep@kit.edu",
    usn="1KT23AI001",
    branch="Artificial Intelligence & Data Science",
    initial_cgpa=9.6
)

print("=== CREATED STUDENT ===")
print("String representation (__str__):\n ", sundeep)
print("\nDeveloper debug representation (__repr__):\n ", repr(sundeep))
print("\nRole Summary:", sundeep.get_role_summary())
print("Monthly Merit Payout:", f"₹{sundeep.calculate_monthly_payout()}")

# Test Encapsulation Guardrail:
print("\n=== TESTING ENCAPSULATION SHIELD ===")
try:
    sundeep.cgpa = 14.5  # Impossible grade!
except ValueError as e:
    print("[BLOCKED BY SETTER]:", e)

# Test Private Attribute Access & Name Mangling:
print("\n=== NAME MANGLING DEMONSTRATION ===")
try:
    print("Trying to access private sundeep.__cgpa directly...")
    print(sundeep.__cgpa)
except AttributeError as e:
    print("[BLOCKED]: Direct access raises AttributeError:", e)

# Behind the scenes, Python mangled the name to _Student__cgpa:
print(f"Accessing via name mangling (sundeep._Student__cgpa): {sundeep._Student__cgpa}")
print("Philosophy: 'We are all consenting adults' — Python trusts you, but warns you not to break the contract!")
````

### 🔬 Deep-Dive: Name Mangling Mechanics & Subclass Collision Prevention

A common beginner misconception is that Python's double-underscore prefix (`__`) is designed to create "private variables" like in Java or C++.

**In reality, Guido van Rossum designed Name Mangling for a much deeper architectural reason: Subclass Collision Prevention in Inheritance Trees.**

#### 1. How Python Textually Mangles Identifiers
Whenever Python encounters an attribute or method starting with at least two underscores (`__`) and at most one trailing underscore, the parser automatically re-writes it behind the scenes:
$$\text{Original: } \texttt{\_\_identifier} \quad \longrightarrow \quad \text{Mangled: } \texttt{\_ClassName\_\_identifier}$$

#### 2. Why Did Python Need This? (The Subclass Collision Problem)
Imagine a large college system where:
- A base class `CollegeMember` maintains an internal security token: `self.__security_token`.
- A subclass `Student` inherits from `CollegeMember` and independently defines its own: `self.__security_token`.

Without Name Mangling, both would write to the same key in `self.__dict__['__security_token']`, and the subclass would **silently overwrite and corrupt** the base class's security token!

With Name Mangling:
- Base class token is stored at: `self._CollegeMember__security_token`
- Subclass token is stored at: `self._Student__security_token`
- **Both live harmoniously in the same instance memory without colliding!**

#### 3. It Applies to Private Methods Too!
Private helper methods (e.g., `def __calculate_disciplinary_deduction(self):`) are also mangled to `_ClassName__calculate_disciplinary_deduction()` so that subclasses cannot accidentally override critical internal parent algorithms.

Let's prove this live by inspecting `instance.__dict__` and demonstrating collision prevention!

````python
# =============================================================================
# LIVE DEMO: NAME MANGLING MECHANICS & SUBCLASS COLLISION SHIELD
# =============================================================================

# 1. Inspect sundeep's actual internal memory dictionary:
print("=== 1. INSPECTING sundeep.__dict__ KEYS ===")
for key, val in sundeep.__dict__.items():
    print(f"  Key: {key:<25} | Value: {val}")

# Notice: There is NO key named '__cgpa'!
# Python textually converted it into: '_Student__cgpa'
print(f"\nDirect access via mangled key: sundeep._Student__cgpa = {sundeep._Student__cgpa}")

# 2. Private Methods Name Mangling Demo:
class SecureCampusMember:
    def __init__(self, name: str):
        self.name = name
        self.__audit_pin = 4321  # Mangled to _SecureCampusMember__audit_pin
    
    # Private method:
    def __generate_encryption_key(self) -> str:
        return f"ENC_{self.name.upper()}_{self.__audit_pin * 7}"
    
    # Public method calling private method internally:
    def get_public_auth_token(self) -> str:
        return f"AUTH_BEARER_{self.__generate_encryption_key()}"

member = SecureCampusMember("Principal_Office")
print("\n=== 2. PRIVATE METHOD MANGLING ===")
print("Public Auth Token:", member.get_public_auth_token())

# Trying to call member.__generate_encryption_key() directly fails:
try:
    member.__generate_encryption_key()
except AttributeError as e:
    print("[BLOCKED]: Calling member.__generate_encryption_key() raises AttributeError:", e)

# But calling via mangled name works:
print("Calling via mangled name:", member._SecureCampusMember__generate_encryption_key())

# 3. Subclass Collision Prevention Proof:
class ParentCampusSystem:
    def __init__(self):
        self.__system_id = "SYS_PARENT_V1"  # -> _ParentCampusSystem__system_id

    def get_parent_id(self):
        return self.__system_id

class ChildStudentSystem(ParentCampusSystem):
    def __init__(self):
        super().__init__()
        self.__system_id = "SYS_STUDENT_V2" # -> _ChildStudentSystem__system_id

    def get_student_id(self):
        return self.__system_id

sys_obj = ChildStudentSystem()
print("\n=== 3. SUBCLASS COLLISION SHIELD DEMO ===")
print("Parent System ID:", sys_obj.get_parent_id())   # Still returns 'SYS_PARENT_V1'!
print("Student System ID:", sys_obj.get_student_id())  # Returns 'SYS_STUDENT_V2'!

print("\nInternal sys_obj.__dict__ hashmap:")
for k, v in sys_obj.__dict__.items():
    print(f"  {k} => {v}")
print("\n[RESULT]: Both parent and child kept their separate variables without overwriting each other!")
````

### 🧬 Concept 4: The 5 Types of Inheritance in KIT College

Let's implement all 5 types of inheritance with real college members:
1. **Single Inheritance**: `CollegeMember` $\rightarrow$ `Student` (one parent $\rightarrow$ one child).
2. **Multilevel Inheritance**: `CollegeMember` $\rightarrow$ `Staff` $\rightarrow$ `TeachingStaff` / `Professor`.
3. **Hierarchical Inheritance**: `CollegeMember` branches into `Student`, `Staff`, `ManagementStaff`.
4. **Multiple Inheritance**: `TeachingAssistant` inherits from BOTH `Student` AND `TeachingStaff`.
5. **Hybrid Inheritance (The Diamond Problem)**: `TeachingAssistant` connects back to `CollegeMember` through two distinct paths!

Let's build this hierarchy:


````python
# 1. Intermediate Parent: Staff (Multilevel step)
class Staff(CollegeMember):
    """Base class for all employed workers at KIT College."""
    def __init__(self, name: str, email: str, emp_id: str, department: str, base_salary: float):
        CollegeMember.__init__(self, name=name, email=email, campus_id=emp_id)
        self.emp_id = emp_id
        self.department = department
        self.__base_salary = base_salary

    @property
    def salary(self) -> float:
        return self.__base_salary

    def calculate_monthly_payout(self) -> float:
        return self.__base_salary

# 2. Multilevel Inheritance: Staff -> TeachingStaff (Professor)
class TeachingStaff(Staff):
    """Professors and Lecturers conducting classes and research."""
    def __init__(self, name: str, email: str, emp_id: str, department: str, base_salary: float, specialization: str):
        Staff.__init__(self, name=name, email=email, emp_id=emp_id, department=department, base_salary=base_salary)
        self.specialization = specialization
        self.assigned_classes: list[str] = []

    def get_role_summary(self) -> str:
        return f"Professor of {self.specialization} [{self.department}] (Emp ID: {self.emp_id})"

    def conduct_lecture(self, topic: str):
        print(f"[LECTURE IN PROGRESS] Prof. {self.name} is teaching '{topic}' to AI batch.")

    def __str__(self) -> str:
        return f"👨‍🏫 Prof. {self.name} | Dept: {self.department} | Specialization: {self.specialization}"

# 3. Hierarchical Child 1: ManagementStaff (Campus leadership)
class ManagementStaff(Staff):
    """Leadership running college administration: Principal, Deans, Registrar."""
    def __init__(self, name: str, email: str, emp_id: str, designation: str, base_salary: float):
        Staff.__init__(self, name=name, email=email, emp_id=emp_id, department="Campus Administration", base_salary=base_salary)
        self.designation = designation  # e.g., 'Principal', 'Dean of Academics'

    def get_role_summary(self) -> str:
        return f"Executive Management: {self.designation} (Emp ID: {self.emp_id})"

    def approve_budget(self, department: str, amount: float):
        print(f"[MANAGEMENT APPROVAL] {self.designation} {self.name} approved budget of ₹{amount:,.2f} for {department}!")

    def __str__(self) -> str:
        return f"🏛️ Executive: {self.name} ({self.designation})"

# 4. Hierarchical Child 2: SupportStaff (Department workers, technicians, lab attendants)
class SupportStaff(Staff):
    """Lab attendants, server administrators, and department technical workers."""
    def __init__(self, name: str, email: str, emp_id: str, department: str, role_title: str, base_salary: float):
        Staff.__init__(self, name=name, email=email, emp_id=emp_id, department=department, base_salary=base_salary)
        self.role_title = role_title  # e.g., 'AI Lab Attendant', 'Systems Administrator'

    def get_role_summary(self) -> str:
        return f"Support Staff: {self.role_title} [{self.department}]"

    def perform_maintenance(self, equipment: str):
        print(f"[MAINTENANCE] {self.name} ({self.role_title}) checked and serviced: {equipment}.")

    def __str__(self) -> str:
        return f"🔧 Support: {self.name} | Role: {self.role_title} [{self.department}]"

# 5. Multiple & Hybrid Inheritance: TeachingAssistant (Student + TeachingStaff)
class TeachingAssistant(Student, TeachingStaff):
    """A senior student who both studies and teaches/grades lab sessions!"""
    def __init__(self, name: str, email: str, usn: str, emp_id: str, branch: str, specialization: str, stipend: float, cgpa: float):
        # Initialize Student parent:
        Student.__init__(self, name=name, email=email, usn=usn, branch=branch, initial_cgpa=cgpa)
        # Initialize TeachingStaff parent:
        TeachingStaff.__init__(self, name=name, email=email, emp_id=emp_id, department=branch, base_salary=stipend, specialization=specialization)

    def get_role_summary(self) -> str:
        return f"Teaching Assistant [USN: {self.usn} / Emp ID: {self.emp_id}] in {self.specialization} (CGPA: {self.cgpa})"

    def calculate_monthly_payout(self) -> float:
        # TA gets monthly stipend + student merit if top grade!
        return self.salary + (1000.0 if self.cgpa >= 9.0 else 0.0)

    def __str__(self) -> str:
        return f"🧑‍💻 TA: {self.name} | USN: {self.usn} | Specialization: {self.specialization} | CGPA: {self.cgpa}"

print("=== ALL 5 INHERITANCE TYPES DEFINED SUCCESSFULLY! ===" )
````

### 🔍 Concept 5: MRO (Method Resolution Order) & C3 Linearization

In multiple and hybrid inheritance, what happens when a method like `get_role_summary()` or `__init__()` exists in multiple parent classes? Which one gets called?

**The Diamond Problem**:
```
        CollegeMember
        /           \
    Student         Staff
        |             |
        |        TeachingStaff
        \           /
       TeachingAssistant
```

Python solves this definitively using the **C3 Linearization Algorithm**:
1. Children precede their parents.
2. Multiple base parents are evaluated in the exact left-to-right order declared in `class Child(Parent1, Parent2):`.
3. No class is visited twice.

Let's inspect the MRO of our `TeachingAssistant` class!


````python
print("=== METHOD RESOLUTION ORDER (MRO) FOR TEACHING ASSISTANT ===")
print("1. Tuple attribute (TeachingAssistant.__mro__):")
for idx, cls in enumerate(TeachingAssistant.__mro__, 1):
    print(f"   Step {idx}: {cls.__name__}")

print("\n2. List method (TeachingAssistant.mro()):")
print([cls.__name__ for cls in TeachingAssistant.mro()])

# Let's create an instance of TeachingAssistant:
ta_meera = TeachingAssistant(
    name="Meera Nair",
    email="meera.ta@kit.edu",
    usn="1KT21AI015",
    emp_id="TA_AI_04",
    branch="Artificial Intelligence & Data Science",
    specialization="Computer Vision & OpenCV",
    stipend=12000.0,
    cgpa=9.7
)

print("\n=== CREATED TEACHING ASSISTANT ===")
print("TA Display:", ta_meera)
print("Role:", ta_meera.get_role_summary())
print(f"Total Monthly Earnings: ₹{ta_meera.calculate_monthly_payout():,.2f}")
ta_meera.conduct_lecture("YOLOv8 Object Detection Lab")
````

### 🎭 Concept 7: Polymorphism (Dynamic, Static & Operator Overloading)

**1. Dynamic Polymorphism (Duck Typing & Runtime Method Overriding)**:
Python doesn't care about rigid type hierarchies; if an object implements `get_role_summary()`, Python calls it! We can iterate through a mixed list of students, professors, executives, and support workers in a single unified loop.

**2. Static Polymorphism: Why No Traditional Method Overloading in Python?**
In C++ or Java, you can write:
```java
void registerStudent(String usn);
void registerStudent(String usn, String hostelRoom);
```
In Python, **defining a second method with the same name OVERWRITES the first one**!
Instead, Python solves method overloading elegantly using:
- Default argument values (`def register(usn, hostel=None):`)
- Variable arguments (`*args`, `**kwargs`)
- `@functools.singledispatch`

**3. Operator Overloading**:
We can teach standard Python math and comparison operators (`+`, `>`, `<`, `==`) how to interact with our custom objects using dunder methods:
- `__add__`: Combines academic marks or credit points (`sundeep + meera`).
- `__gt__` / `__lt__`: Enables direct sorting by CGPA (`sundeep > rahul`).
- `__eq__`: Compares whether two student objects share the identical USN.

Let's implement and test this!


````python
# Adding Operator Overloading to Student:
class AcademicStudent(Student):
    """Student with Operator Overloading for +, >, <, and ==."""
    
    def __init__(self, name: str, email: str, usn: str, branch: str, cgpa: float, exam_marks: int):
        super().__init__(name, email, usn, branch, cgpa)
        self.exam_marks = exam_marks

    # 1. Operator Overloading: + (Combines total exam marks of study partners)
    def __add__(self, other):
        if isinstance(other, AcademicStudent):
            return self.exam_marks + other.exam_marks
        elif isinstance(other, (int, float)):
            return self.exam_marks + other
        return NotImplemented

    # 2. Operator Overloading: > (Ranks students by CGPA)
    def __gt__(self, other):
        if isinstance(other, AcademicStudent):
            return self.cgpa > other.cgpa
        return NotImplemented

    # 3. Operator Overloading: < (Enables sorted() out of the box)
    def __lt__(self, other):
        if isinstance(other, AcademicStudent):
            return self.cgpa < other.cgpa
        return NotImplemented

    # 4. Operator Overloading: == (Compares unique USN)
    def __eq__(self, other):
        if isinstance(other, AcademicStudent):
            return self.usn == other.usn
        return False

    # 5. The Hash Contract: __hash__ (Required when overriding __eq__!)
    def __hash__(self):
        """Two students with the identical USN produce the identical hash code."""
        return hash(self.usn)

# Create Academic Students:
s1 = AcademicStudent("Sundeep", "sundeep@kit.edu", "1KT23AI001", "AI-DS", cgpa=9.6, exam_marks=98)
s2 = AcademicStudent("Arjun Das", "arjun@kit.edu", "1KT23AI002", "AI-DS", cgpa=8.9, exam_marks=86)
s3 = AcademicStudent("Priya Sharma", "priya@kit.edu", "1KT23AI003", "AI-DS", cgpa=9.8, exam_marks=99)
s_duplicate = AcademicStudent("Sundeep Duplicate Entry", "sundeep2@kit.edu", "1KT23AI001", "AI-DS", cgpa=9.6, exam_marks=98)

print("=== OPERATOR OVERLOADING IN ACTION ===")
# 1. Test +
combined_score = s1 + s2
print(f"s1 + s2 (Combined Project Marks): {s1.exam_marks} + {s2.exam_marks} = {combined_score}")

# 2. Test > and <
print(f"Is Sundeep higher CGPA than Arjun? (s1 > s2): {s1 > s2}")
print(f"Is Priya higher CGPA than Sundeep? (s3 > s1): {s3 > s1}")

# 3. Automatic Sorting with sorted():
roster = [s1, s2, s3]
print("\nSorted Students by CGPA (Ascending):")
for stud in sorted(roster):
    print(f"  {stud.name}: CGPA {stud.cgpa}")

# 4. Test == (Comparing USN uniqueness)
print(f"\nDo s1 and s2 have identical USN? (s1 == s2): {s1 == s2}")
print(f"Do s1 and s_duplicate have identical USN? (s1 == s_duplicate): {s1 == s_duplicate}")

print("\n=== DYNAMIC POLYMORPHISM IN A SINGLE LOOP ===")
campus_roster = [s1, prof_anand, principal, lab_tech, ta_meera]
for member in campus_roster:
    # Notice: No if isinstance(...) checks needed! Pure duck typing!
    print(f"• [{member.__class__.__name__}] -> {member.get_role_summary()} | Monthly: ₹{member.calculate_monthly_payout():,.2f}")
````

### 🔬 Deep-Dive: String Internals, Immutability & The Hash Contract (`__hash__` & `__eq__`)

Have you ever wondered why strings in Python are **immutable**, how `hash()` works behind the scenes, and why dictionaries can look up keys in instantaneous $\mathcal{O}(1)$ time?

![String Internals & Character Pooling Architecture](../../images/day2_string_internals_and_memory_pooling.png)

#### 1. The Hash Contract in Python
Python enforces a strict mathematical rule for hash tables:
$$\mathbf{a == b \implies \text{hash}(a) == \text{hash}(b)}$$
> **The Golden Trap**: In Python, as soon as a class defines `__eq__`, **Python automatically sets `__hash__ = None`**!
> Python does this defensively: if two objects can be equal, but you don't define how their hash is calculated, storing them in a `set` or `dict` could violate the hash contract.
> To make our `AcademicStudent` instances hashable (so we can store them in a `set` or use them as `dict` keys), we **must explicitly implement `__hash__`** matching `self.usn`!



#### 2. The "Ganesh" & "Kavya" Case: CPython 1-Character Singleton Pooling
When you evaluate `'Ganesh'[1]` and `'Kavya'[1]`, does Python allocate new memory for the letter `'a'`?
**NO!**
In CPython, all **256 Latin-1 single characters (ASCII 0 to 255)** are pre-allocated at startup as global static singletons (`characters[256]`).
- The `'a'` in `"Ganesh"` points directly to the pre-allocated singleton `'a'`.
- The `'a'` at index 1 in `"Kavya"` points to the **exact same singleton `'a'`**!
- The `'a'` at index 4 in `"Kavya"` **also** points to the **exact same singleton `'a'`**!
- **Proof in Python**: `('Ganesh'[1] is 'Kavya'[1])` returns `True`!

#### 3. String Internals in CPython (PEP 393 Flexible String Representation)
Inside the CPython C-runtime, strings are represented by the `PyASCIIObject` / `PyCompactUnicodeObject` struct:
- **1 Byte per char** (Latin-1/ASCII): For strings like `"1KT23AI001"` or `"Sundeep"`. Highly memory efficient.
- **2 Bytes per char** (UCS-2): For European accented characters, Cyrillic, Greek.
- **4 Bytes per char** (UCS-4): For emojis (`🎓`, `🤖`, `⚡`) and Asian scripts.
- **`ob_shash`**: In C, every string struct stores a precomputed 64-bit integer hash. It is calculated **once** on demand and cached forever!

#### 4. Whole-String Interning (`sys.intern()`)
Python automatically **interns** string literals that look like Python identifiers (e.g. variable names, function names, dictionary keys like `"name"`, `"usn"`):
- Interned strings share the **exact same memory address in heap RAM** (`a is b` returns `True`).
- This allows Python's dictionary lookups (`student.__dict__['usn']`) to compare **raw C memory pointer addresses in a single CPU clock cycle**, completely bypassing character-by-character string comparisons!

#### 5. Why MUST Strings Be Immutable?
1. **Hash Table Integrity**: A dictionary bucket index is calculated as `hash(key) & (capacity - 1)`. If strings were mutable, altering `student.usn` in-place would change its hash. The key would become **permanently lost** inside the wrong bucket, destroying the dictionary!
2. **Thread Safety**: Immutable strings can be read concurrently across hundreds of worker threads without mutex locking.
3. **Security**: Network socket URLs, database connection strings, and file paths cannot be maliciously altered by downstream functions.

#### 6. How Hashing Works: SipHash-2-4 & The HashDoS Defense
Python calculates string hashes using the cryptographically hardened **SipHash-2-4** algorithm.
Every time you launch Python, it generates a random 128-bit secret seed (`PYTHONHASHSEED`). This prevents **HashDoS attacks**, where malicious actors send crafted form inputs that all collide into the same bucket to crash servers with $\mathcal{O}(N)$ CPU denial-of-service!

Let's test the Hash Contract, set deduplication, string interning, and hash consistency live:

````python
# =============================================================================
# LIVE DEMO: STRING INTERNALS, HASH CONTRACT & SET DEDUPLICATION
# =============================================================================
import sys

# 1. Test Hash Contract on AcademicStudent:
print("=== 1. TESTING THE HASH CONTRACT ON STUDENTS ===")
print(f"hash(s1): {hash(s1)} (Based on USN '{s1.usn}')")
print(f"hash(s_duplicate): {hash(s_duplicate)} (Based on USN '{s_duplicate.usn}')")
print(f"Are s1 and s_duplicate equal? -> {s1 == s_duplicate}")
print(f"Do they share identical hash? -> {hash(s1) == hash(s_duplicate)}")

# 2. Storing Custom Objects in a set() (Automatic Deduplication!):
print("\n=== 2. SET DEDUPLICATION VIA __hash__ AND __eq__ ===")
student_set = {s1, s2, s3, s_duplicate}  # s_duplicate has same USN as s1
print(f"Total items added to set: 4")
print(f"Unique students surviving in set: {len(student_set)}")
for stud in student_set:
    print(f"  • {stud.name} [USN: {stud.usn}] -> Hash: {hash(stud)}")

# 3. Using Custom Objects as Dictionary Keys:
print("\n=== 3. USING STUDENTS AS DICTIONARY KEYS ===")
scholarship_allocations = {
    s1: "₹25,000 AI Research Grant",
    s3: "₹30,000 Director's Gold Merit Medal"
}
print(f"Lookup scholarship by object s1: {scholarship_allocations[s1]}")
# Lookup works even with duplicate object because hash & equality match:
print(f"Lookup scholarship by s_duplicate: {scholarship_allocations[s_duplicate]}")

# 4. String Interning Demonstration (sys.intern):
print("\n=== 4. STRING INTERNING IN CPYTHON ===")
str_a = "KIT_CAMPUS_2026"
str_b = "KIT_CAMPUS_2026"
print(f"str_a == str_b (Value equality)   : {str_a == str_b}")
print(f"str_a is str_b (Pointer identity)  : {str_a is str_b} (Exact same memory address: {hex(id(str_a))})")

# Dynamically created strings that bypass compile-time interning:
dyn_a = "".join(["kit_", "autonomous_", "token"])
dyn_b = "".join(["kit_", "autonomous_", "token"])
print(f"dyn_a == dyn_b (Value match)       : {dyn_a == dyn_b}")
print(f"dyn_a is dyn_b (Before sys.intern) : {dyn_a is dyn_b} (Different heap addresses!)")

# Explicitly interning dynamic strings:
interned_a = sys.intern(dyn_a)
interned_b = sys.intern(dyn_b)
print(f"interned_a is interned_b (After)   : {interned_a is interned_b} (Merged to single heap address: {hex(id(interned_a))})")


# 4B. The "Ganesh" & "Kavya" 1-Character Singleton Pooling Experiment:
print("\n=== 4B. THE 'GANESH' & 'KAVYA' CHARACTER POOLING EXPERIMENT ===")
w1 = "Ganesh"
w2 = "Kavya"
print(f"Does 'Ganesh'[1] share the exact memory pointer with 'Kavya'[1]? -> {w1[1] is w2[1]}")
print(f"Memory address of 'Ganesh'[1] ('a'): {hex(id(w1[1]))}")
print(f"Memory address of 'Kavya'[1]  ('a'): {hex(id(w2[1]))}")
print(f"Memory address of 'Kavya'[4]  ('a'): {hex(id(w2[4]))}")
print(f"Memory address of literal 'a'      : {hex(id('a'))}")
print("Proof: All occurrences of 'a' resolve to the SAME singleton in RAM!")

# 5. Immutability & Hash Collision Defense:
print("\n=== 5. WHY STRINGS MUST BE IMMUTABLE ===")
print("String Hash:", hash(s1.usn))
print("Attempting to modify a string character in-place:")
try:
    s1.usn[0] = 'X'
except TypeError as e:
    print("  [PROTECTED BY PYTHON]:", e)
print("Because strings are immutable, their hash never changes in memory, guaranteeing O(1) dictionary safety!")
````

### 📦 Concept 8: Magic Dunder Methods & The Department Container

In Python, classes become truly first-class citizens when they integrate seamlessly with Python's syntax:
- `len(dept)` $\rightarrow$ triggers `__len__`
- `dept[0]` or `dept["1KT23AI001"]` $\rightarrow$ triggers `__getitem__`
- `"1KT23AI001" in dept` $\rightarrow$ triggers `__contains__`
- `for s in dept:` $\rightarrow$ triggers `__iter__`

Let's build our `KITDepartment` class!


````python
class KITDepartment:
    """Container class managing students and faculty in a department at KIT."""
    
    def __init__(self, dept_name: str, dept_code: str):
        self.dept_name = dept_name
        self.dept_code = dept_code
        self._students: list[AcademicStudent] = []
        self._usn_index: dict[str, AcademicStudent] = {}

    def admit_student(self, student: AcademicStudent):
        """Adds student to roster and registers their USN in our fast hashmap."""
        if student.usn in self._usn_index:
            raise ValueError(f"Student with USN '{student.usn}' already enrolled!")
        self._students.append(student)
        self._usn_index[student.usn] = student
        print(f"[ADMISSION CONFIRMED] Enrolled {student.name} ({student.usn}) in Dept of {self.dept_name}.")

    # --- DUNDER MAGIC METHODS ---
    def __len__(self) -> int:
        """Allows writing len(department) directly!"""
        return len(self._students)

    def __getitem__(self, key):
        """Allows bracket indexing by integer index OR by USN string!"""
        if isinstance(key, int):
            return self._students[key]
        elif isinstance(key, str):
            return self._usn_index[key]
        raise TypeError(f"Invalid key type: {type(key)}. Use int index or str USN.")

    def __contains__(self, usn_or_student) -> bool:
        """Allows writing '1KT23AI001' in department!"""
        if isinstance(usn_or_student, str):
            return usn_or_student in self._usn_index
        elif isinstance(usn_or_student, AcademicStudent):
            return usn_or_student.usn in self._usn_index
        return False

    def __iter__(self):
        """Allows writing 'for student in department:'"""
        return iter(self._students)

    def __str__(self) -> str:
        return f"<KIT Department: {self.dept_name} [{self.dept_code}] | Total Enrolled: {len(self)} students>"

# Let's instantiate and test our Department container:
ai_dept = KITDepartment("Artificial Intelligence & Data Science", "AI-DS")

ai_dept.admit_student(s1)
ai_dept.admit_student(s2)
ai_dept.admit_student(s3)

print("\n=== TESTING DEPARTMENT DUNDER METHODS ===")
print("Department:", ai_dept)
print(f"Total Enrolled (via len()): {len(ai_dept)}")

# Test indexing by integer:
print(f"First student (ai_dept[0]): {ai_dept[0].name}")

# Test indexing by USN string:
print(f"Lookup by USN (ai_dept['1KT23AI003']): {ai_dept['1KT23AI003'].name} (CGPA: {ai_dept['1KT23AI003'].cgpa})")

# Test membership check with 'in':
print(f"Is '1KT23AI001' in ai_dept?: {'1KT23AI001' in ai_dept}")
print(f"Is '1KT23EC099' in ai_dept?: {'1KT23EC099' in ai_dept}")

# Test iteration:
print("\nIterating over department with for loop:")
for student in ai_dept:
    print(f"  • {student.usn} -> {student.name} ({student.branch})")
````

### ⚡ Concept 9: Why Python is Most Efficient & The Power of AI Libraries

Why is Python the undisputed world leader in Artificial Intelligence, Machine Learning, and Data Science?

![Why Python Rules AI](../../images/day2_python_efficiency_and_libraries.png)

1. **Developer Velocity > CPU Cycles**:
   - In 2026, engineering payroll is expensive; cloud GPU compute is cheap. Writing clean, readable code in 10 lines of Python instead of 100 lines of C++ saves weeks of development time.
2. **The "C Engine Under the Hood" Secret**:
   - Python itself is written in C (**CPython**).
   - High-performance libraries like **NumPy**, **PyTorch**, **TensorFlow**, and **Pandas** are NOT running heavy mathematical tensor loops in pure Python!
   - They execute compiled **C, C++, Fortran, and NVIDIA CUDA** kernels underneath.
   - Python serves as the **elegant, readable steering wheel** controlling a high-octane C++ Formula 1 engine!

Let's prove this with a live speed benchmark comparing pure Python vs NumPy:


````python
import time
import numpy as np

# Let's compare computing the sum of squares of 1,000,000 student score records:
N = 1_000_000

# 1. Pure Python Loop:
start_py = time.perf_counter()
total_py = sum(i ** 2 for i in range(N))
py_duration = (time.perf_counter() - start_py) * 1000

# 2. Vectorized NumPy (C Engine):
start_np = time.perf_counter()
arr = np.arange(N, dtype=np.int64)
total_np = np.sum(arr ** 2)
np_duration = (time.perf_counter() - start_np) * 1000

print(f"=== PERFORMANCE BENCHMARK (N = {N:,} elements) ===")
print(f"• Pure Python Loop Duration : {py_duration:.2f} ms")
print(f"• NumPy (C Engine) Duration : {np_duration:.2f} ms")
print(f"⚡ NumPy is {py_duration / np_duration:.1f}x faster because it runs vectorized C code under the hood!")
print("\nConclusion: Python combines maximum human development speed with C-level raw execution speed.")
````

````python
# Write your challenge solution below:
class LabTechnician(SupportStaff):
    def __init__(self, name: str, email: str, emp_id: str, department: str, base_salary: float, lab_assigned: str):
        super().__init__(name, email, emp_id, department, "Lead Lab Technician", base_salary)
        self.lab_assigned = lab_assigned

    def run_ai_benchmark(self, gpu_name: str):
        print(f"[BENCHMARK RUNNING] {self.name} is stress-testing {gpu_name} in {self.lab_assigned}...")
        print(f"  Result: 142 TFLOPS recorded. GPU temperatures normal at 62°C! ✅")

# Test your class:
my_tech = LabTechnician(
    name="Sundeep",
    email="sundeep.tech@kit.edu",
    emp_id="TECH_AI_09",
    department="Artificial Intelligence & Data Science",
    base_salary=55000.0,
    lab_assigned="KIT Generative AI Supercomputing Lab"
)

print(my_tech)
print("Role:", my_tech.get_role_summary())
my_tech.run_ai_benchmark("NVIDIA RTX 4090 (24GB VRAM)")

print("\nInheritance Path (MRO):")
for step, cls in enumerate(LabTechnician.mro(), 1):
    print(f"  Step {step}: {cls.__name__}")
````
